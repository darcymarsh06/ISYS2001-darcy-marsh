```python
def budget_tracker():
    print("=== 📊 Weekly Expense & Budget Tracker ===")

    # 1. Ask for weekly income
    while True:
        try:
            income = float(input("Enter your total weekly income: $"))
            if income < 0:
                print("Income cannot be negative. Please try again.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a numerical value.")

    # 2. Collect weekly expenses
    expenses = {}
    print("\nEnter your weekly expenses. Type 'done' when you are finished.")

    while True:
        item = input("Expense name (or 'done' to finish): ").strip()
        if item.lower() == 'done':
            break
        if not item:
            continue

        while True:
            try:
                amount = float(input(f"  Enter cost for '{item}': $"))
                if amount < 0:
                    print("  Amount cannot be negative. Try again.")
                    continue
                expenses[item] = amount
                break
            except ValueError:
                print("  Please enter a valid number for the amount.")

    # 3. Calculate total expenses and savings
    total_expenses = sum(expenses.values())
    savings = income - total_expenses

    # Display weekly summary
    print("\n" + "=" * 40)
    print("        WEEKLY BUDGET SUMMARY         ")
    print("=" * 40)
    print(f"Total Income:    ${income:,.2f}")
    print(f"Total Expenses:  ${total_expenses:,.2f}")

    if expenses:
        print("\nBreakdown of expenses:")
        for name, amt in expenses.items():
            print(f"  • {name}: ${amt:,.2f}")

    print("-" * 40)
    if savings >= 0:
        print(f"Net Savings:     ${savings:,.2f}")
    else:
        print(f"Current Deficit: -${abs(savings):,.2f}")
    print("=" * 40)

    # 4. Check additional extra expense
    print("\n--- 🛒 Extra Expense Decision Check ---")
    while True:
        try:
            extra_expense = float(input("Enter the cost of the extra item you want to buy: $"))
            if extra_expense < 0:
                print("Expense cannot be negative.")
                continue
            break
        except ValueError:
            print("Please enter a valid number.")

    # 5. Evaluate if expense is within budget
    remaining_balance = savings - extra_expense

    print("\n--- Decision Result ---")
    if extra_expense <= savings:
        print(f"✅ **Approved!** You can proceed with this purchase.")
        print(f"Remaining savings after purchase: ${remaining_balance:,.2f}")
    else:
        deficit = extra_expense - savings
        print(f"❌ **Over Budget!** This expense exceeds your available savings.")
        print(f"You need an extra ${deficit:,.2f} to make this purchase without going into debt.")

if __name__ == "__main__":
    budget_tracker()
```
